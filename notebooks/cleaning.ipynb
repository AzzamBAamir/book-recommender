```python
# import the dataset
# what does it look like?

books = pd.read_csv("/Users/azzambinaamir/book-recommender/data/Books.csv",",")
ratings = pd.read_csv("/Users/azzambinaamir/book-recommender/data/Ratings.csv",",")
users = pd.read_csv("/Users/azzambinaamir/book-recommender/data/Users.csv",",")
```

```python
# regex cleaning

# Function to clean a single string column
def clean_string_column(column: pd.Series, pattern: str = r'[^a-zA-Z]') -> pd.Series:
    """
    Clean a string column by removing all non-English characters (anything except a-zA-Z).
    Args:
    - column (pd.Series): The string column to clean.
    - pattern (str): The regex pattern to use for cleaning (default is to remove non-English characters).
    
    Returns:
    - pd.Series: The cleaned column.
    """
    return column.str.replace(pattern, '', regex=True)

# Function to clean all string columns in a DataFrame
def clean_dataframe(df: pd.DataFrame) -> pd.DataFrame:
    """
    Clean all string columns in a DataFrame by removing non-English characters.
    Args:
    - df (pd.DataFrame): The DataFrame to clean.
    
    Returns:
    - pd.DataFrame: The cleaned DataFrame.
    """
    for col in df.select_dtypes(include='object').columns:
        df[col] = clean_string_column(df[col])
    return df
```

```python
items_to_clean = [books, ratings, users]

for i in items_to_clean:
    i = clean_dataframe(i)
```
